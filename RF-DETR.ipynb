```python
# Imports
from pathlib import Path
from rfdetr import RFDETRMedium
```

```python
model = RFDETRMedium()
```

```python
# Dtaaset and Output path
from roboflow import Roboflow
from getpass import getpass

ROBOFLOW_API_KEY = getpass("Enter your Roboflow API Key: ")

rf = Roboflow(api_key=ROBOFLOW_API_KEY)

project = (
    rf.workspace("afzals-workspace-wqtvl")
      .project("eyelidtumordetection")
)

version = project.version(1)

dataset = version.download("coco")

# Dataset path
DATASET_DIR = Path(dataset.location)

# Output Path
OUTPUT_DIR = Path(
    "/content/drive/MyDrive/RF_DETR_Output_V2"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

assert DATASET_DIR.exists(), f"Dataset not found: {DATASET_DIR}"

print(DATASET_DIR)
print(OUTPUT_DIR)
```

```python
# Train
model.train(
    dataset_dir=str(DATASET_DIR),
    output_dir=str(OUTPUT_DIR),

    epochs=20,

    batch_size=4,
    grad_accum_steps=4,

    lr=1e-4,
)
```
