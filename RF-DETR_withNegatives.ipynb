This notebook trains RF-DETR Medium using annotated eyelid-tumour images together with normal images containing no tumour annotations. The experiment investigates whether explicit negative samples improve false-positive control and model generalisation.

## Workflow

1. Mount Google Drive.
2. Install RF-DETR and the required dependencies.
3. Initialise the RF-DETR Medium model.
4. Download the COCO-format dataset containing tumour annotations and normal-image negatives.
5. Configure the dataset and output directories.
6. Train the model and resume training from a saved checkpoint when required.
7. Save the model checkpoints and validation metrics.

## Experimental setup

- Positive images: eyelid images with tumour bounding-box annotations
- Negative images: normal images without tumour bounding boxes
- Detection target: eyelid tumour
- Annotation format: COCO
- Training/validation split: created in Roboflow

## Outputs

- Best regular and EMA checkpoints
- Resumable training checkpoints
- Training logs
- Validation metrics

```python
# Imports
from pathlib import Path
from rfdetr import RFDETRMedium
```

```python
model = RFDETRMedium()
```

```python
# Dtaaset and Output path
from roboflow import Roboflow
from getpass import getpass

ROBOFLOW_API_KEY = getpass("Enter your Roboflow API Key: ")

rf = Roboflow(api_key=ROBOFLOW_API_KEY)

project = (
    rf.workspace("salini-s-iiitk-s-workspace")
      .project("eyelid-lesion-detection-with-neg")
)

version = project.version(1)

dataset = version.download("coco")

# Dataset path
DATASET_DIR = Path(dataset.location)

# Output Path
OUTPUT_DIR = Path(
    "/content/drive/MyDrive/RF_DETR_Output_WithNegatives"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

assert DATASET_DIR.exists(), f"Dataset not found: {DATASET_DIR}"

print(DATASET_DIR)
print(OUTPUT_DIR)
```

```python
# Train
model.train(
    dataset_dir=str(DATASET_DIR),
    output_dir=str(OUTPUT_DIR),

    epochs=20,

    batch_size=4,
    grad_accum_steps=4,

    lr=1e-4,
)
```

When code stops running for train, execute from below

```python
# Checking checkpoint saved
from pathlib import Path

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/RF_DETR_Output_WithNegatives"
)

print("Files available in the output folder:\n")

for file_path in sorted(OUTPUT_DIR.iterdir()):
    size_mb = file_path.stat().st_size / (1024 ** 2)
    print(f"{file_path.name:45s} {size_mb:.2f} MB")
```

```python
# Slecting correct resumable cehckpoint
from pathlib import Path

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/RF_DETR_Output_WithNegatives"
)

# Check the most likely exact-resume checkpoint names first
candidate_checkpoints = [
    OUTPUT_DIR / "latest.ckpt",
    OUTPUT_DIR / "last.ckpt",
    OUTPUT_DIR / "checkpoint.pth",
]

RESUME_CHECKPOINT = next(
    (path for path in candidate_checkpoints if path.exists()),
    None
)

# If interval checkpoints exist, use the most recently modified one
if RESUME_CHECKPOINT is None:
    interval_checkpoints = list(
        OUTPUT_DIR.glob("checkpoint_interval_*.ckpt")
    )

    if interval_checkpoints:
        RESUME_CHECKPOINT = max(
            interval_checkpoints,
            key=lambda path: path.stat().st_mtime
        )

if RESUME_CHECKPOINT is None:
    raise FileNotFoundError(
        "No full resumable checkpoint was found.\n"
        "checkpoint_best_regular.pth and checkpoint_best_ema.pth "
        "cannot provide an exact optimizer/epoch resume."
    )

print("Training will resume from:")
print(RESUME_CHECKPOINT)
```

```python

model.train(
    dataset_dir=str(DATASET_DIR),

    # Continue saving in the same Google Drive folder
    output_dir=str(OUTPUT_DIR),

    #  total target is 20 epochs.
    # (if epoch 9 completed, it continues with epoch 10).
    epochs=20,

    batch_size=4,
    grad_accum_steps=4,
    lr=1e-4,

    # Exact continuation from the saved training state
    resume=str(RESUME_CHECKPOINT),
)
```
